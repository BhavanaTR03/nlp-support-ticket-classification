```python
# 1. IMPORT LIBRARIES
import pandas as pd
import re
import nltk

from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score, confusion_matrix, ConfusionMatrixDisplay

import matplotlib.pyplot as plt

# Download NLTK resources (run once)
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
# Load the CSV file 
df = pd.read_csv("cw_data.csv", header=None)
df.columns = ["text", "label"]

# Verify class distribution
print(df['label'].value_counts())
```

```python
# 3. DOMAIN-SPECIFIC CLEANING
abbrev_dict = {
    "pls": "please",
    "acct": "account",
    "pw": "password",
    "cant": "cannot",
    "u": "you",
    "msg": "message"
}

def clean_abbreviations(text):
    tokens = text.split()
    tokens = [abbrev_dict.get(word.lower(), word) for word in tokens]
    return " ".join(tokens)
```

```python
# 4. TEXT PREPROCESSING
lemmatizer = WordNetLemmatizer()

def clean_text_advanced(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = clean_abbreviations(text)
    tokens = word_tokenize(text)
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return " ".join(tokens)

df['clean_text'] = df['text'].apply(clean_text_advanced)
```

```python
# 5. FEATURE ENGINEERING (TF-IDF)
tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1,2),
    sublinear_tf=True,
    stop_words='english'
)
```

```python
# 6. TRAIN-TEST SPLIT
X_train_text, X_test_text, y_train, y_test = train_test_split(
    df['clean_text'],
    df['label'],
    test_size=0.2,
    stratify=df['label'],
    random_state=42
)
```

```python
# Vectorize after split to prevent data leakage
X_train = tfidf_vectorizer.fit_transform(X_train_text)
X_test = tfidf_vectorizer.transform(X_test_text)
```

```python
# 7. MODEL + GRIDSEARCH
lr = LogisticRegression(class_weight='balanced', max_iter=1000)

param_grid = {
    'C': [0.01, 0.1, 1, 10]
}

grid_lr = GridSearchCV(
    lr,
    param_grid,
    scoring='f1_macro',
    cv=5,
    n_jobs=-1
)

grid_lr.fit(X_train, y_train)

best_lr = grid_lr.best_estimator_

print("Best Parameters:", grid_lr.best_params_)
```

```python
# 8. EVALUATION
y_pred = best_lr.predict(X_test)

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

f1 = f1_score(y_test, y_pred, average='macro')
print("Macro F1 Score:", f1)
```

```python
#9. CONFUSION MATRIX

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Create confusion matrix FIRST
cm = confusion_matrix(y_test, y_pred, labels=best_lr.classes_)

# Then plot
plt.figure(figsize=(12,10))

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=best_lr.classes_)
disp.plot(cmap='Blues', xticks_rotation=90)

plt.title("Confusion Matrix - Pipeline 2 (Tuned Logistic Regression)")
plt.tight_layout()

plt.show()
```

```python

errors_df = pd.DataFrame({
    'text': df.loc[y_test.index, 'text'],
    'actual': y_test,
    'predicted': y_pred
})

errors_df = errors_df[errors_df['actual'] != errors_df['predicted']]
errors_df.sample(5)
```

```python
# SECOND MODEL: NAIVE BAYES

from sklearn.naive_bayes import MultinomialNB

nb = MultinomialNB()

param_grid_nb = {
    'alpha': [0.1, 0.5, 1.0]
}

grid_nb = GridSearchCV(
    nb,
    param_grid_nb,
    scoring='f1_macro',
    cv=5,
    n_jobs=-1
)

# Train
grid_nb.fit(X_train, y_train)

# Best model
best_nb = grid_nb.best_estimator_

print("Best NB Parameters:", grid_nb.best_params_)

# Predict
y_pred_nb = best_nb.predict(X_test)

# Evaluate
from sklearn.metrics import f1_score

f1_nb = f1_score(y_test, y_pred_nb, average='macro')
print("Naive Bayes Macro F1:", f1_nb)
```
